```python
import matplotlib.pyplot as plt
from main_funcs import *
import numpy as np
import os
```

```python
base_dirs = {
    "Without": Path("data/Without EDM"),
    "With": Path("data/With EDM"),
    "Without New": Path("data/Without EDM NEW"),
    "With New": Path("data/With EDM NEW")
}

data_all = load_array_data(base_dirs)
```

```python
folder = "Weight_Maps2/"
i=0

weight_dict = {}

theta_vals = data_all["With"][0]["Y"][:,0]

for filename in os.listdir(folder):
    i+=1
    if not filename.endswith(".npz"):
        continue

    npz = np.load(os.path.join(folder, filename), allow_pickle=True)
    lam = npz["lam"]
    col_strength = np.sum(np.abs(npz["weight"]), axis=0)
    j = int(np.argmax(col_strength))
    w = npz["weight"][:, j]
    weight_dict[float(lam)] = w
```

```python
plt.plot(theta_vals, weight_dict[0.250])
plt.ylabel(r"w($\theta,\lambda)$")

plt.show()
```
